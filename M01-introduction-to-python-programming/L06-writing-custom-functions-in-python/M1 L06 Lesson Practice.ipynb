# Writing Custom Functions: Reusable Order Calculations

In this practice you will write your own functions to calculate order totals, apply discounts and describe orders. You will use positional, keyword and default arguments, `*args` and `**kwargs`, and see how variable scope works.

## Background

The retailer calculates order totals, discounts and VAT many times a day. Writing each calculation once as a function, and calling it whenever it is needed, keeps the code short and makes every calculation consistent.

## Data Description

All the practices in this module use **Retail Orders.csv**: 200 orders from 2025 placed with a small online retailer based in Ireland. The data is fictional and was created for teaching.

## 1. Load the Dataset

We import pandas and read the CSV file into a DataFrame. Keep **Retail Orders.csv** in the same folder as this notebook.

```python
import pandas as pd

df = pd.read_csv("Retail Orders.csv")
df.head()
```

## 2. Define and Call a Simple Function

`def` defines a function. Nothing happens until the function is called by name with brackets.

```python
def print_title():
    print("Retail Orders Report")

print_title()
```

## 3. Arguments and return

`order_total()` takes a quantity and a unit price and **returns** the total. Because it returns a value, we can store the result and use it again.

```python
def order_total(quantity, unit_price):
    return quantity * unit_price

t = order_total(3, 45.0)
print(t)
print(t + order_total(5, 189.0))
```

A function that only prints gives nothing back. Its result is `None`:

```python
def show_total(quantity, unit_price):
    print(quantity * unit_price)

result = show_total(3, 45.0)
print(result)
```

## 4. Positional and Keyword Arguments

Positional arguments are matched by order. Keyword arguments are matched by name, so their order does not matter.

```python
print(order_total(2, 24.99))
print(order_total(unit_price=24.99, quantity=2))
```

## 5. Default Arguments

`discount` defaults to 0, so it can be left out. Parameters with defaults come after those without.

```python
def apply_discount(total, discount=0):
    return round(total * (1 - discount), 2)

print(apply_discount(745.0))
print(apply_discount(745.0, 0.10))
print(apply_discount(745.0, discount=0.25))
```

## 6. Use the Functions on the Dataset

The same function works for every order. Here we apply it to the first five orders.

```python
for i in range(5):
    q = int(df["quantity"][i])
    p = float(df["unit_price"][i])
    print(df["order_id"][i], order_total(q, p))
```

## 7. Any Number of Values with *args

`*totals` collects any number of positional arguments into a tuple, so one function can add up two orders or ten.

```python
def revenue(*totals):
    return sum(totals)

print(revenue(135.0, 13.5))
print(revenue(135.0, 13.5, 945.0, 600.0))
```

## 8. Named Details with **kwargs

`**details` collects any number of keyword arguments into a dictionary.

```python
def describe_order(**details):
    for key, value in details.items():
        print(key, ":", value)

describe_order(order_id="ORD-1004", city="Cork", product="Ergonomic Chair", quantity=5)
```

## 9. Local and Global Scope

`VAT_RATE` is created outside any function, so it is **global** and every function can read it. `vat` is created inside the function, so it is **local** and disappears when the function ends.

```python
VAT_RATE = 0.23

def add_vat(net):
    vat = net * VAT_RATE
    return round(net + vat, 2)

print(add_vat(100))
```

Assigning to a name inside a function creates a new local variable. The global value is unchanged:

```python
def test_rate():
    VAT_RATE = 0.135
    print("Inside function:", VAT_RATE)

test_rate()
print("Outside function:", VAT_RATE)
```

## 10. Validation and Nested Functions

A function can check its input before calculating. Here a nested helper function does the multiplication, and the outer function rejects negative quantities.

```python
def safe_total(quantity, unit_price):
    def multiply(a, b):
        return a * b

    if quantity < 0:
        print("Error: quantity cannot be negative.")
    else:
        return multiply(quantity, unit_price)

print(safe_total(4, 39.5))
print(safe_total(-2, 39.5))
```
