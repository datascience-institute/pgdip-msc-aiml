# Lists, Tuples and Dictionaries: Organising Order Data

In this practice you will store order information in Python's three main collection types. You will use a list for a customer's basket, a tuple for fixed values, and a dictionary for a single order record.

## Background

The retailer's customer service team often needs to look up an order, adjust a basket or check delivery details. Choosing the right collection type makes each of these tasks simpler.

## Data Description

All the practices in this module use **Retail Orders.csv**: 200 orders from 2025 placed with a small online retailer based in Ireland. The data is fictional and was created for teaching.

| Column Name | Description |
|---|---|
| order_id | Unique order identifier |
| order_date | Date the order was placed (YYYY-MM-DD) |
| delivery_date | Date the order was delivered; blank if not yet delivered |
| customer_name | Customer's name, as typed into the order system |
| city | Delivery city |
| product | Product ordered |
| category | Product category: Electronics, Office or Home |
| quantity | Number of units ordered |
| unit_price | Price of one unit, in euro |
| priority | Order priority: Urgent, Normal or Minor |
| status | Order status: Delivered, Shipped or Cancelled |

## 1. Load the Dataset

We import pandas and read the CSV file into a DataFrame. Keep **Retail Orders.csv** in the same folder as this notebook.

```python
import pandas as pd

df = pd.read_csv("Retail Orders.csv")
df.head()
```

## 2. Create a List of Products

A list is an ordered, changeable sequence in square brackets. We take the products from the first five orders. `len()` counts the items.

```python
basket = list(df["product"][0:5])
print(basket)
print(len(basket))
```

## 3. Access Items by Position

Indexing starts at 0. A slice `[1:3]` returns positions 1 and 2, and a negative index counts from the end.

```python
print(basket[0])
print(basket[1:3])
print(basket[-1])
```

## 4. Change the List

Lists are mutable, so we can update, add and remove items.

```python
basket[2] = "Desk Lamp"             # update the item at index 2
basket.append("Kettle")             # add to the end
basket.insert(1, "USB-C Hub")       # add at index 1
print(basket)
```

```python
del basket[1]                       # remove by position
basket.remove("Kettle")             # remove by value
print(basket)
```

## 5. Search and Sort a List

`index()` finds an item's position, `in` checks whether it exists, and `count()` counts how often it appears.

```python
print(basket.index("Desk Lamp"))
print("Whiteboard" in basket)
print(basket.count("A4 Notebook"))
```

```python
prices = list(df["unit_price"][0:6])
prices.sort()
print(prices)

prices.sort(reverse=True)
print(prices)
```

## 6. Tuples for Fixed Values

A tuple is like a list but cannot be changed. The retailer always delivers to the same five cities, so a tuple suits them.

```python
delivery_cities = ("Dublin", "Cork", "Galway", "Limerick", "Waterford")
print(delivery_cities[0])
print(delivery_cities[1:3])
```

Trying to change a tuple raises a `TypeError`.

The next cell is meant to fail. If you use *Run All*, Jupyter stops here, so run the cells after it one at a time.

```python
delivery_cities[0] = "Sligo"
```

To extend a tuple, we create a new one by joining two tuples:

```python
expanded_cities = delivery_cities + ("Sligo",)
expanded_cities
```

## 7. A Dictionary for One Order

A dictionary stores **key: value** pairs, so each value has a name. We describe order ORD-1001.

```python
order = {
    "order_id": df["order_id"][0],
    "product": df["product"][0],
    "quantity": int(df["quantity"][0]),
    "unit_price": float(df["unit_price"][0]),
    "status": df["status"][0],
}
order
```

## 8. Read Values from the Dictionary

Square brackets read a value by its key. `get()` does the same but returns a default instead of an error when the key is missing.

```python
print(order["product"])
print(order.get("city", "Not recorded"))
print(order.keys())
print(order.values())
```

## 9. Update the Dictionary

We change the quantity, add the delivery city, and remove the status.

```python
order["quantity"] = 4
order["city"] = df["city"][0]
del order["status"]
order
```

```python
order.update({"priority": "Urgent", "quantity": 5})
removed = order.pop("priority")
print(removed)
order
```

## 10. Nested Structures

A list of dictionaries holds several records with the same fields, which is how a table of orders looks inside Python.

```python
orders = [
    {"order_id": "ORD-1002", "city": "Dublin", "items": [3, 45.0]},
    {"order_id": "ORD-1003", "city": "Cork", "items": [3, 4.5]},
]
print(orders[1]["city"])
print(orders[0]["items"][1])
```

- searched and sorted a list
- used a tuple for values that should not change, and saw the error when you tried to change one
- stored an order as a dictionary and read, added, changed and removed its values
