## 1. Print a Welcome Message

`print()` displays text on the screen. Text must be enclosed in quotes.

A line that starts with `#` is a **comment**. Python ignores it, so you can use comments to explain what your code does.

```python
# Print a title for the report
print("Daily Order Report")
```

`print()` also works with numbers and with simple calculations. Numbers do not need quotes.

```python
print(200)
print(3 * 45)
```

## 2. Create Objects for the First Order

An **object** stores a value so you can use it again later. The `=` operator assigns a value to an object.

Here we store the details of order ORD-1002.

```python
# Details of order ORD-1002
order_id_1 = "ORD-1002"
quantity_1 = 3
unit_price_1 = 45.0
```

Typing an object's name on its own displays its value, just as it would in the Spyder console.

```python
quantity_1
```

```python
unit_price_1
```

## 3. Calculate the Order Total

The `*` operator multiplies. An order's total is the quantity multiplied by the unit price. We store the result in a new object, `total_1`, so we can use it later.

```python
# Order total = quantity x unit price
total_1 = quantity_1 * unit_price_1
total_1
```

## 4. Calculate the Other Two Orders

We repeat the same steps for orders ORD-1003 and ORD-1004.

```python
# Details of order ORD-1003
order_id_2 = "ORD-1003"
quantity_2 = 3
unit_price_2 = 4.5
total_2 = quantity_2 * unit_price_2

# Details of order ORD-1004
order_id_3 = "ORD-1004"
quantity_3 = 5
unit_price_3 = 189.0
total_3 = quantity_3 * unit_price_3

print(total_2)
print(total_3)
```

## 5. Add and Subtract Order Totals

The `+` operator adds and the `-` operator subtracts. Adding the three totals gives the revenue for these orders.

```python
# Revenue from the three orders
revenue = total_1 + total_2 + total_3
revenue
```

Subtracting shows how far apart the largest and smallest orders are. Here the largest is ORD-1004 and the smallest is ORD-1003.

```python
# Gap between the largest and smallest order
gap = total_3 - total_2
gap
```

## 6. Check the Order of Operations

When a calculation mixes operators, Python multiplies before it adds, just as in ordinary arithmetic.

Suppose delivery costs a fixed €5.00 plus €2.50 for each item. For order ORD-1002, with 3 items, the delivery charge is 5 + 3 × 2.50 = €12.50. Python gets this right without any extra effort:

```python
# Fixed fee plus a charge per item
delivery_1 = 5 + quantity_1 * 2.5
delivery_1
```

Brackets tell Python to do the part inside them first. If we put brackets around the addition, Python adds before it multiplies, which here gives the wrong charge of €20.00:

```python
# Brackets change the order: (5 + 3) x 2.50
wrong_delivery_1 = (5 + quantity_1) * 2.5
wrong_delivery_1
```

## 7. Print Results with Commas

To print several items on one line, separate them with commas. `print()` puts a space between each item, and you can mix text and numbers freely.

```python
print("Order", order_id_1, "total:", total_1)
print("Order", order_id_2, "total:", total_2)
print("Order", order_id_3, "total:", total_3)
print("Revenue for the three orders:", revenue)
```

## 8. Join Text with the + Operator

Used with text, `+` joins two strings into a single string. It does not add a space, so we include the spaces ourselves.

```python
city_1 = "Dublin"
product_1 = "Whiteboard"

# Join text to build a delivery label
label_1 = product_1 + " for delivery to " + city_1
print(label_1)
```

## 9. Format a Message with the % Operator

The `%` operator inserts values into a message. `%s` is replaced by a string and `%d` by a whole number. The values go in brackets after the `%`, in the same order as the placeholders.

```python
print("Order %s has %d items" % (order_id_3, quantity_3))
```
