# Data Types in Python: Exploring the Retail Orders Data

In this practice you will load the retail orders dataset and look at the data types Python uses to store it: numbers, text, categories and True/False values. You will then build vectors, a matrix, an array, a data frame and a list from the orders.

## Background

The operations team at a small Irish online retailer has exported its 2025 orders to a CSV file. Before anyone analyses the data, they want to know what kind of values each column holds, because Python treats numbers, text and categories differently.

## Data Description

All the practices in this module use **Retail Orders.csv**: 200 orders from 2025 placed with a small online retailer based in Ireland. The data is fictional and was created for teaching.

| Column Name | Description |
|---|---|
| order_id | Unique order identifier |
| order_date | Date the order was placed (YYYY-MM-DD) |
| delivery_date | Date the order was delivered; blank if not yet delivered |
| customer_name | Customer's name, as typed into the order system |
| city | Delivery city |
| product | Product ordered |
| category | Product category: Electronics, Office or Home |
| quantity | Number of units ordered |
| unit_price | Price of one unit, in euro |
| priority | Order priority: Urgent, Normal or Minor |
| status | Order status: Delivered, Shipped or Cancelled |

## 1. Load the Dataset

`pd.read_csv()` reads a CSV file into a pandas DataFrame. This is a new function: it takes the file name in quotes and returns the whole table. Keep **Retail Orders.csv** in the same folder as this notebook.

`df.head()` shows the first five rows.

```python
import pandas as pd

df = pd.read_csv("Retail Orders.csv")
df.head()
```

## 2. Check the Structure with info()

`info()` lists every column with its data type. `int64` columns hold whole numbers, `float64` columns hold decimals, and text columns are stored as strings (shown here as `str`; older versions of pandas show `object`).

```python
df.info()
```

## 3. Numeric Types: Integer and Float

To look at single values, we take one value from a column. `df["quantity"][0]` means *the `quantity` column, first row* (Python counts from 0).

`type()` tells us the data type of an object.

```python
quantity = int(df["quantity"][0])
unit_price = float(df["unit_price"][0])

print(quantity, type(quantity))
print(unit_price, type(unit_price))
```

Converting a float to an integer with `int()` **truncates** the decimal part. It does not round. Here we use a price of €24.99 from the Wireless Mouse:

```python
mouse_price = 24.99
print(int(mouse_price))
print(type(int(mouse_price)))
```

## 4. Character Type

Text is stored as a string (`str`). A number in quotes is text, not a number. `str()` converts any object into a string.

```python
order_id = df["order_id"][0]
print(order_id, type(order_id))

quantity_text = str(quantity)
print(quantity_text, type(quantity_text))
```

## 5. Categorical Type

`priority` can only take one of three values. A **categorical** object stores a fixed set of possible values, called **levels**. We convert the column with `pd.Categorical()` and check its levels with `unique()`.

```python
priority = pd.Categorical(df["priority"])
priority.unique()
```

By default the levels are in alphabetical order. The `categories=` argument sets the order we want, from least to most urgent:

```python
priority = pd.Categorical(df["priority"], categories=["Minor", "Normal", "Urgent"])
priority.categories
```

## 6. Logical Type

A logical value is either `True` or `False`. `isinstance()` checks whether an object is of a given type, and a comparison asks a True-or-False question.

```python
print(isinstance(unit_price, int))
print(isinstance(unit_price, float))

is_large_order = quantity > 2
print(is_large_order, type(is_large_order))
```

## 7. Vectors

A vector is a sequence of values of the same type. Here we build a numeric vector and a character vector from the first six orders. `list()` turns a column extract into a plain Python list.

```python
quantities = list(df["quantity"][0:6])
cities = list(df["city"][0:6])

print(quantities)
print(cities)
```

With a NumPy array, one comparison checks every value at once and gives a **logical vector**:

```python
import numpy as np

q = np.array(quantities)
q > 2
```

## 8. Matrix

A matrix is two-dimensional. `reshape(3, 2)` arranges the six quantities into 3 rows and 2 columns, filled row by row.

```python
m = np.array(quantities).reshape(3, 2)
m
```

Putting the matrix into a DataFrame lets us name its rows and columns:

```python
named = pd.DataFrame(m, index=["Pair 1", "Pair 2", "Pair 3"], columns=["First", "Second"])
named
```

## 9. Array

An array can have more than two dimensions. `np.arange(1, 13)` creates the numbers 1 to 12, one for each month of the year. `reshape(4, 3)` arranges them as the four quarters, with three months each.

```python
months = np.arange(1, 13).reshape(4, 3)
months
```

## 10. Data Frame and List

A data frame can hold different types in each column. Here we build a small one from three vectors: product names, quantities and prices.

```python
products = list(df["product"][0:3])
qty = list(df["quantity"][0:3])
price = list(df["unit_price"][0:3])

small = pd.DataFrame({"product": products, "quantity": qty, "unit_price": price})
small
```

```python
small.info()
```

Unlike a vector, a list can mix types. `list(a + b)` joins two lists into one:

```python
mixed = list(qty + products)
mixed
```
