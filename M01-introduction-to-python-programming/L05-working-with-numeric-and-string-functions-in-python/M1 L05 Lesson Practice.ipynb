# Numeric and String Functions: Calculating and Cleaning Orders

In this practice you will use Python's built-in numeric functions and operators to work with order values. You will then use string functions to tidy customer names and build readable labels.

## Background

The finance team needs order values rounded and summarised. The customer service team has noticed that some customer names were typed in inconsistent capitals, such as `JACK DOYLE` or `aoife murphy`. Both problems are solved with Python's built-in functions.

## 1. Load the Dataset

We import pandas and read the CSV file into a DataFrame. Keep **Retail Orders.csv** in the same folder as this notebook.

```python
import pandas as pd

df = pd.read_csv("Retail Orders.csv")
df.head()
```

## 2. Rounding Functions

Order ORD-1006 is for 5 pairs of headphones at €149.00. Suppose a 7% discount is applied to it. The discounted value has many decimal places, and each rounding function treats it differently.

```python
import math

value = 5 * 149.00 * 0.93
print(value)
print(round(value, 2))      # nearest value with 2 decimals
print(math.ceil(value))     # up to the next whole number
print(math.floor(value))    # down to the whole number below
print(math.trunc(value))    # drop the decimal part
```

## 3. Absolute Value, Square Root, Log and Exponential

`abs()` removes the sign, which is useful for the size of a difference. The customer's budget was €700, so the discounted order is €7.15 under budget, shown as a negative difference. `math.sqrt()`, `math.log10()` and `math.exp()` are also common in data analysis.

```python
budget = 700
difference = round(value - budget, 2)
print(difference)
print(abs(difference))

print(math.sqrt(144))
print(math.log10(1000))
print(round(math.exp(1), 4))
```

## 4. Arithmetic Operators

Notebooks are sold in packs of 4. For an order of 11 notebooks, `//` gives the number of full packs and `%` gives the number left over. `**` raises a number to a power.

```python
notebooks = 11
pack_size = 4

print(notebooks / pack_size)    # division
print(notebooks // pack_size)   # full packs
print(notebooks % pack_size)    # notebooks left over
print(2 ** 3)                   # power
```

## 5. Relational and Membership Operators

Relational operators compare values and return `True` or `False`. The `in` operator checks whether a value belongs to a sequence.

```python
quantity = int(df["quantity"][0])
print(quantity, quantity >= 5, quantity != 3, quantity == 1)

print("Cork" in ["Dublin", "Cork", "Galway"])
print(3 in range(1, 6))
print(6 in range(1, 6))     # the upper limit of range() is not included
```

## 6. Change Case with upper(), lower() and casefold()

Some customer names were typed entirely in capitals or entirely in lower case. A name typed all in capitals is unchanged by `upper()`, and a name typed all in lower case is unchanged by `lower()`, so comparing each name with those versions picks them out.

```python
names = list(df["customer_name"])
messy = [name for name in names if name == name.upper() or name == name.lower()]
print(len(messy))
print(messy[0:5])
```

```python
print([x.lower() for x in messy[0:3]])
print([x.upper() for x in messy[0:3]])
print("SÍOBHÁN BYRNE".casefold())
```

## 7. Split and Join Strings

`split()` breaks a string into parts and `join()` puts a list of strings back together with a separator.

```python
name = df["customer_name"][0]
parts = name.split()
print(parts)

print(", ".join([parts[1], parts[0]]))      # surname first
```

```python
date_parts = df["order_date"][0].split("-")
print(date_parts)
print("/".join([date_parts[2], date_parts[1], date_parts[0]]))
```

## 8. Extract and Replace Substrings

Square brackets extract part of a string. `replace()` swaps every instance of one substring for another.

```python
order_id = df["order_id"][0]
print(order_id[0:3])
print(order_id[4:8])

print(order_id.replace("ORD", "INV"))
```

Shortening city names to their first three letters is a common way to make compact labels:

```python
cities = list(df["city"][0:6])
short = [city.replace(city, city[0:3]) for city in cities]
short
```

## 9. Concatenate and Format Strings

`+` joins strings, `%` inserts values into a message, and `format()` does the same with `{}` placeholders.

```python
product = df["product"][0]
city = df["city"][0]
price = df["unit_price"][0]

print(product + " to " + city)
print("Order %s has %d items" % (order_id, quantity))
print("{} costs {} per unit".format(product, price))
print("Deliver {1} to {0}".format(city, product))
```
